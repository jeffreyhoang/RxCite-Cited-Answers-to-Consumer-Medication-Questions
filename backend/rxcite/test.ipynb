# 1. Download one label straight from openFDA

```python
import json, urllib.parse, urllib.request

query = urllib.parse.urlencode(
    {
        "search": 'openfda.generic_name:"warfarin"',
        "sort": "effective_time:desc",
        "limit": 1,
    }
)
resp = json.load(urllib.request.urlopen(f"https://api.fda.gov/drug/label.json?{query}"))
label = resp["results"][0]
```

# 2. Print label sections

```python
metadata = ["set_id", "id", "version", "effective_time", "openfda"]

for i, l in enumerate(label):
    print(i, l) if l not in metadata else None
```
